```python
import os
from pathlib import Path
from pprint import pprint
import json
from dotenv import load_dotenv
import urllib.request
from landingai_ade import LandingAIADE
import logging
```

```python
load_dotenv()
```

```python
##logginh
logging.basicConfig(filename='example.log', level=logging.DEBUG,
format='%(asctime)s %(levelname)s:%(message)s')
```

```python
#env variable
LANDINGAI_API_KEY = os.getenv('VISION_AGENT_API_KEY')
```

```python
#setup client
try:
    logging.info('Setting up client')
    client = LandingAIADE(apikey=LANDINGAI_API_KEY)
    logging.info('Client setup successful!!')
except Exception as e:
    logging.info(f'client did not start successfully: {e}')
```

```python
#define fucntion to dowenload documents

def download_doc(pdf_url:str, dest_folder:str):
    try:
        urllib.request.urlretrieve(pdf_url,dest_folder)
    except Exception as e:
        logging.info('Download unsuccessful')
    logging.info(f'Document downloaded to: {dest_folder}')

download_doc('https://docs.landing.ai/examples/bank-statement.pdf',
             'C:\\Users\\LENOVO\\Desktop\\healthcare-assistant-prototype\\HC-Assistant-Prototype\\data\\sample_docs\\sample.pdf')
```

```python
#parsing
response = client.parse(
    document = Path("C:\\Users\\LENOVO\\Desktop\\healthcare-assistant-prototype\\HC-Assistant-Prototype\\data\\sample_docs\\sample.pdf"),
    model = "dpt-2-latest",
    split="page"
)
logging.info(f"Extracted {len(response.chunks)} chunks")
```

```python
#verify chunking
for i, chunk in enumerate(response.chunks[:3]):
    print(f"\n--- Chunk {i+1} ---")
    print(f"Type: {chunk.type}")
    if chunk.grounding:
        print(f"Page: {chunk.grounding.page}")
    print(f"Content: {chunk.markdown[:150]}...")
```
